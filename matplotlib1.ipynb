```python
import matplotlib.pyplot as plt
```

```python
x=[2,4,6,8,10]
y=[1,3,5,7,9]
y1=[1,2,3,4,5]
y2=[4,8,1,9,10]
y3=[5,9,2,3,7]
plt.plot(x,y)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x,y1)
plt.show()
```

```python
plt.subplot(2,2,3)
plt.plot(x,y2)

plt.subplot(2,2,4)
plt.plot(x,y3)
plt.show()
```
