```python
import numpy as np
arr=np.array(list(map(float,input().split())))
arr=arr[arr>=0]
print(round(np.mean(arr),1))
```

```python
import numpy as np
arr=np.array(list(map(int,input().split())))
threshold=int(input())
filtered=arr[arr>threshold]
if len(filtered)==0:
    print(0.0)
else:
    print(round(np.mean(filtered),2))
```
