```python
import matplotlib.pyplot as plt
```

```python
x=[2,4,6,8,10]
y=[1,3,4,5,6]
y1=[5,6,7,8,9]
plt.plot(x,y)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x,y1)
plt.show()
```

```python
x=[1,2,4,6,7,9]
y=[2,3,8,7,10,11]
y1=[6,7,9,10,11,12]
y2=[4,6,8,1,2,3]
y3=[9,11,2,9,10,13]

plt.plot(x,y)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x,y1)
plt.show()
```

```python
plt.subplot(2,2,3)
plt.plot(x,y2)

plt.subplot(2,2,4)
plt.plot(x,y3)
plt.show()
```
